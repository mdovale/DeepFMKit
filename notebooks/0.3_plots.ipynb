```python
from deepfmkit.core import DeepFrame
from deepfmkit import physics
import numpy as np
from deepfmkit.plotting import default_rc
from matplotlib.ticker import FuncFormatter
import matplotlib.pyplot as plt

plt.rcParams.update(default_rc)
```

```python
dff = DeepFrame(raw_file="../tests/raw_data.gz", raw_labels=["raw1"])
dff.fit("raw1", n=20)
```

```python
axes = dff.plot()
plt.show()
```

```python
axs = dff.plot(which=["phi", "m"], figsize=(5, 2), lw=1, c="tomato")
plt.show()
```

```python
ax = dff.raws["raw1"].plot(figsize=(5, 3), xrange=[0, 0.01], c="lime")
ax.grid(False)
plt.show()
```

```python

cycles = 1
f_samp = 200e3
fm = 1e3
m = 7.2

laser = physics.LaserConfig()
laser.fm = fm

ifo = physics.IfoConfig()
ifo.ref_arml = 0.15
ifo.meas_arml = 0.2
ifo.phi = 0.0

main = physics.SimConfig(
    label="Hello", laser_config=laser, ifo_config=ifo, f_samp=f_samp
)

laser.set_df_for_effect(ifo, m)

laser.psi = np.pi / 3
ifo.phi = np.pi / 3

ax = main.plot_harmonics(N=10)
plt.show()
```

```python
ax = main.plot_harmonics_vs_phi(N=10)
plt.show()
```

```python
sg = physics.SignalGenerator()

fig, ax = plt.subplots(figsize=(3.375, 1.8), dpi=150)
raw = sg.generate(main, n_seconds=cycles / fm, mode="snr", snr_db=100)["main"]
ax = raw.plot(ax=ax, ls="-", lw=1.0)

ax.set_ylabel("Voltage (a.u.)")
ax.xaxis.set_major_formatter(FuncFormatter(lambda x, _: f"{x * 1e3:.1f}"))
ax.set_xlabel("Time (ms)")
ax.grid(False)

plt.show()
```
